```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/Projetar-Solucoes/materiais-trilha/main/semana-04/solicitacoes_semana4.csv")
display (df)
```

```python
df = df.drop_duplicates()
display(df)
```

```python

df_copy = df.copy()

df_copy["categoria"] = df_copy["categoria"].str.strip().str.lower()

padrao = {
    "suporte ti": "Suporte TI",
    "suporte de ti": "Suporte TI",
    "acesso": "Acesso",
    "financeiro": "Financeiro",
    "infraestrutura": "Infraestrutura",
    "infra": "Infraestrutura",
    "dúvida": "Dúvida",
    "duvida": "Dúvida",
}

df_copy["categoria"] = df_copy["categoria"].map(padrao)

display(df_copy["categoria"].value_counts())
```

```python
import pandas as pd

df_copy["status"] = df_copy["status"].str.capitalize()

df_copy["avaliacao"] = df_copy["avaliacao"].replace(7, float("nan"))

media_avaliacao = df_copy.groupby("categoria")["avaliacao"].mean()

display(media_avaliacao)

display(df_copy["categoria"].value_counts())
```
